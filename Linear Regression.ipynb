```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
california_data = fetch_california_housing(as_frame=True)
df = california_data.frame
```

```python
x =df.drop("MedHouseVal", axis=1)
y = df["MedHouseVal"]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(x)
```

```python

X_scaled = pd.DataFrame(
    X_scaled,
    columns=x.columns
)
X_scaled.head()
```

### 2. Regression Algorithm Implementation 

● Implement the following regression algorithms:
○ Linear Regression
○ Decision Tree Regressor
○ Random Forest Regressor
○ Gradient Boosting Regressor
○ Support Vector Regressor (SVR)

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)
```

### 1  Linear Regression

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
y_pred_linear = linear_model.predict(X_test)
```

### 2️ Decision Tree Regressor

```python
from sklearn.tree import DecisionTreeRegressor
tree_model = DecisionTreeRegressor(random_state=42)
tree_model.fit(X_train, y_train)
y_pred_tree = tree_model.predict(X_test)
```

### 3️ Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor
forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
forest_model.fit(X_train, y_train)
y_pred_forest = forest_model.predict(X_test)
```

### 4️ Gradient Boosting Regressor

```python
from sklearn.ensemble import GradientBoostingRegressor
gradient_model = GradientBoostingRegressor(
    random_state=42
)
gradient_model.fit(X_train, y_train)
y_pred_gradient = gradient_model.predict(X_test)
```

### 5️ Support Vector Regressor (SVR)

```python
from sklearn.svm import SVR
svr_model = SVR()
svr_model.fit(X_train, y_train)
y_pred_svr = svr_model.predict(X_test)
```

# 3. Model Evaluation and Comparison

### ● Evaluate the performance of each algorithm using the following metrics:
○ Mean Squared Error (MSE)
○ Mean Absolute Error (MAE)
○ R-squared Score (R2)

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
mse_linear = mean_squared_error(y_test, y_pred_linear)
mae_linear = mean_absolute_error(y_test, y_pred_linear)
r2_linear = r2_score(y_test, y_pred_linear)
print("Linear Regression Results")
print("MSE:", mse_linear)
print("MAE:", mae_linear)
print("R² Score:", r2_linear)
```

```python
mse_tree = mean_squared_error(y_test, y_pred_tree)
mae_tree = mean_absolute_error(y_test, y_pred_tree)
r2_tree = r2_score(y_test, y_pred_tree)
print("Decision Tree Results")
print("MSE:", mse_tree)
print("MAE:", mae_tree)
print("R² Score:", r2_tree)
```

```python
mse_forest = mean_squared_error(y_test, y_pred_forest)
mae_forest = mean_absolute_error(y_test, y_pred_forest)
r2_forest = r2_score(y_test, y_pred_forest)
print("Random Forest Results")
print("MSE:", mse_forest)
print("MAE:", mae_forest)
print("R² Score:", r2_forest)
```

```python
mse_gradient = mean_squared_error(y_test, y_pred_gradient)
mae_gradient = mean_absolute_error(y_test, y_pred_gradient)
r2_gradient = r2_score(y_test, y_pred_gradient)
print("Gradient Boosting Results")
print("MSE:", mse_gradient)
print("MAE:", mae_gradient)
print("R² Score:", r2_gradient)
```

```python
mse_svr = mean_squared_error(y_test, y_pred_svr)
mae_svr = mean_absolute_error(y_test, y_pred_svr)
r2_svr = r2_score(y_test, y_pred_svr)
print("SVR Results")
print("MSE:", mse_svr)
print("MAE:", mae_svr)
print("R² Score:", r2_svr)
```

```python
results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest",
        "Gradient Boosting",
        "SVR"
    ],

    "MSE": [
        mse_linear,
        mse_tree,
        mse_forest,
        mse_gradient,
        mse_svr
    ],

    "MAE": [
        mae_linear,
        mae_tree,
        mae_forest,
        mae_gradient,
        mae_svr
    ],

    "R² Score": [
        r2_linear,
        r2_tree,
        r2_forest,
        r2_gradient,
        r2_svr
    ]
})

print(results)
```

### The best-performing algorithm


```python
best_model = results.loc[results["R² Score"].idxmax()]

print("Best Performing Model:")
print(best_model)
```

### The worst-performing algorithm

```python
worst_model = results.loc[results["R² Score"].idxmin()]

print("Worst Performing Model:")
print(worst_model)
```
