## Banking System Using Python

### Account Login

```python
print ("""● Allow users to log in using an account number/username and PIN/password
● Validate user credentials
● Display appropriate error messages for invalid login attempts""")
```

```python
class Login:
    def __init__(self):
        self.accounts = {
            "2001": {"pin": "1234"},
            "2002": {"pin": "5432"}
        }

    def login(self):
        Account = input("Enter Account Number: ")
        pin = input("Enter PIN: ")
        if Account in self.accounts:
            if self.accounts[Account]["pin"] == pin:
                print("Successful")
            else:
                print("Invalid PIN")
        else:
            print("Account Not Found")


kyc = Login()
kyc.login()
```

## Amount Deposit

```python
print("""● Allow logged-in users to deposit money
● Deposit amount must be greater than zero
● Update and display the current balance after deposit""")
```

```python
class Deposit:

    def __init__(self):
        self.balance = 5000

    def deposit(self):
        amount = float(input("Enter Deposit Amount: "))

        if amount > 0:
            self.balance += amount
            print("Deposit Successful")
            print("Current Balance:", self.balance)
        else:
            print("Deposit amount must be greater than zero.")


kyc = Deposit()
kyc.deposit()
```

## Amount Withdrawal

```python
print("""● Allow users to withdraw money
● Withdrawal amount must be greater than zero""")
```

```python
class Withdraw:

    def __init__(self):
        self.balance = 5000

    def withdraw(self):
        amount = float(input("Enter Withdrawal Amount: "))

        if amount <= 0:
            print("Withdrawal amount must be greater than zero.")

        elif amount > self.balance:
            print("Insufficient Balance!")

        else:
            self.balance -= amount
            print("Withdrawal Successful")
            print("Remaining Balance:", self.balance)


kyc = Withdraw()
kyc.withdraw()
```
