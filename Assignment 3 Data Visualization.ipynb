```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

1. Countplot: Passenger Count by Embarked Port
● Create a countplot to display the number of passengers who boarded from each
port (embarked), separated by sex.

```python
titanic = sns.load_dataset("titanic")
plt.figure(figsize=(8, 6))
sns.countplot(data=titanic, x="embarked", hue="sex", palette="Set2")
plt.title("Passenger Count by Embarked Port (Split by Sex)", fontsize=14, fontweight="bold")
plt.xlabel("Port of Embarkation (C = Cherbourg, Q = Queenstown, S = Southampton)")
plt.ylabel("Number of Passengers")
plt.legend(title="Sex")
plt.tight_layout()
plt.savefig("1_countplot_embarked.png", dpi=150)
plt.show()
```

2. Swarmplot: Age Distribution across Classes
● Create a swarmplot to visualize the distribution of age across passenger classes
(pclass), categorized by sex.

```python
titanic = sns.load_dataset("titanic")

plt.figure(figsize=(9, 6))
sns.swarmplot(data=titanic, x="pclass", y="age", hue="sex" ,dodge=True, size=3, palette="Set1")
plt.title("Age Distribution Across Passenger Classes (by Sex)", fontsize=14, fontweight="bold")
plt.xlabel("Passenger Class")
plt.ylabel("Age (years)")
plt.legend(title="Sex")
plt.tight_layout()

plt.show()

# Observation:
# Median age rises with class: 1st class skews older (30s-40s), 3rd class is
# younger and denser, reflecting immigrant/working-class families. Within each
# class, male and female age spreads look similar.
```

3. Pie Chart: Gender Distribution
● Create a pie chart using Matplotlib to show the percentage of male vs. female
passengers. - Add percentage labels using autopct - Use explode for one slice to
highlight it.

```python

titanic = sns.load_dataset("titanic")

gender_counts = titanic["sex"].value_counts()
colors = ["#4C72B0", "#DD8452"]
explode = [0.08 if s == "male" else 0 for s in gender_counts.index]

plt.figure(figsize=(7, 7))
plt.pie(
    gender_counts.values,
    labels=gender_counts.index.str.capitalize(),
    autopct="%1.1f%%",
    explode=explode,
    colors=colors,
    startangle=90,
    shadow=True,
    textprops={"fontsize": 12}
)
plt.title("Gender Distribution of Titanic Passengers", fontsize=14, fontweight="bold")
plt.axis("equal")
plt.tight_layout()
plt.savefig("3_pie_gender.png", dpi=150)
plt.show()

# Observation:
# About 64.8% of passengers were male and 35.2% female. The male slice is
# exploded to highlight that men made up nearly two-thirds of everyone aboard,
# despite women having a much higher survival rate.
```

4. Heatmap: Correlation Matrix
● Create a heatmap to show correlations between age, fare, pclass, sibsp, parch,
and survived. (Use seaborn.heatmap with annotations)

```python
sns.set_theme(style="whitegrid")
titanic = sns.load_dataset("titanic")

cols = ["age", "fare", "pclass", "sibsp", "parch", "survived"]
corr = titanic[cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", square=True,
            linewidths=0.5, cbar_kws={"label": "Correlation Coefficient"})
plt.title("Correlation Matrix of Key Numerical Features", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("4_heatmap_correlation.png", dpi=150)
plt.show()

# Observation:
# pclass has the strongest negative correlation with survived (-0.34) - lower
# class number (better class) meant better survival odds. fare correlates
# positively with survival (+0.26). age, sibsp, and parch show weak correlations.
```

6. Scatter Plot: Age vs Fare
● Create a scatter plot to explore the relationship between age and fare, colored by
survival status.
● Use a seaborn scatterplot with hue='survived'. Each plot must have titles, labels,
and legends where needed.

```python

titanic = sns.load_dataset("titanic")

plt.figure(figsize=(9, 6))
sns.scatterplot(data=titanic, x="age", y="fare", hue="survived",
                 palette={0: "crimson", 1: "seagreen"}, alpha=0.7, s=60)
plt.title("Age vs Fare, Colored by Survival Status", fontsize=14, fontweight="bold")
plt.xlabel("Age (years)")
plt.ylabel("Fare ($)")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.tight_layout()
plt.savefig("6_scatter_age_fare.png", dpi=150)
plt.show()

# Observation:
# Survivors (green) cluster more toward higher fares, especially above $50,
# reinforcing that wealth/class was protective. Non-survivors (red) are
# concentrated in the lower-fare band across all ages.
```
