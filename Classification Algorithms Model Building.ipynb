```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
X = df.drop("target", axis=1)
y = df["target"]
print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget values:")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
imputer = SimpleImputer(strategy="median")
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)
print("\nPreprocessing completed successfully.")
print("\nShape of scaled training data:", X_train_scaled.shape)
print("Shape of scaled testing data:", X_test_scaled.shape)
```

The breast cancer dataset was loaded using load_breast_cancer(as_frame=True). Missing values were checked, and the dataset contains no missing values. However, SimpleImputer with the median strategy was included to make the preprocessing process robust. The data was split into training and testing sets before scaling. StandardScaler was applied because the features have different numeric ranges, and distance-based or scale-sensitive models such as Logistic Regression, SVM, and k-NN require scaled data.

## Classification Algorithm Implementation


### Logistic Regression

```python
logistic_model = LogisticRegression(max_iter=5000,random_state=42)
```

```python
logistic_model.fit(X_train_scaled, y_train)
y_pred_logistic = logistic_model.predict(X_test_scaled)
logistic_accuracy = accuracy_score(y_test, y_pred_logistic)
print("Logistic Regression Accuracy:", logistic_accuracy)
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_logistic))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_logistic))


```

Logistic Regression is a supervised binary classification algorithm. It predicts the probability that a tumor belongs to the benign or malignant class. It is suitable for this dataset because the target has two classes and the dataset contains numerical medical measurements. The model is trained using scaled training data and evaluated using accuracy, classification report, and confusion matrix.

## Decision Tree Classifier

```python
decision_tree_model = DecisionTreeClassifier(random_state=42)
```

```python
decision_tree_model.fit(X_train_imputed, y_train)
y_pred_tree = decision_tree_model.predict(X_test_imputed)
tree_accuracy = accuracy_score(y_test, y_pred_tree)
print("Decision Tree Accuracy:", tree_accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred_tree))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_tree))
```

A Decision Tree classifies data by repeatedly splitting it based on feature conditions. It is suitable because it can model non-linear relationships and produces understandable decision rules. Feature scaling is not required because Decision Trees use split conditions rather than distance calculations.

## RandomForestClassifier

```python
random_forest_model = RandomForestClassifier(n_estimators=300,random_state=42)
```

```python
random_forest_model.fit(X_train_imputed, y_train)
y_pred_forest = random_forest_model.predict(X_test_imputed)
forest_accuracy = accuracy_score(y_test, y_pred_forest)
print("Random Forest Accuracy:", forest_accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred_forest))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_forest))
```

Random Forest is an ensemble model that combines predictions from many Decision Trees through majority voting. It is suitable for this dataset because it can handle multiple numerical features, capture non-linear patterns, and reduce the overfitting risk of a single Decision Tree.

## Support Vector Machine (SVM)

```python
svm_model = SVC(kernel="rbf",C=1.0,gamma="scale",random_state=42)
```

```python
svm_model.fit(X_train_scaled, y_train)
y_pred_svm = svm_model.predict(X_test_scaled)
svm_accuracy = accuracy_score(y_test, y_pred_svm)
print("SVM Accuracy:", svm_accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred_svm))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_svm))
```

SVM finds the optimal boundary, called a hyperplane, that separates benign and malignant samples with the maximum margin. The RBF kernel allows the model to handle non-linear patterns. SVM is suitable for this numerical binary classification dataset, but feature scaling is essential because SVM is sensitive to feature magnitude.

##  k-Nearest Neighbors (k-NN)

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
```

```python
knn_model.fit(X_train_scaled, y_train)
y_pred_knn = knn_model.predict(X_test_scaled)
knn_accuracy = accuracy_score(y_test, y_pred_knn)
print("k-NN Accuracy:", knn_accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred_knn))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_knn))


```

k-NN predicts the class of a new sample by looking at the classes of its nearest training samples. Here, k=5 means the model uses the five nearest neighbors and predicts the majority class. It is suitable for this small numerical dataset, but scaling is essential because k-NN uses distance calculations.

Compare the performance of the five classification algorithms.
Identify which algorithm performed best and which performed worst.

```python
comparison_df = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "SVM",
        "k-NN"
    ],

    "Accuracy": [
        accuracy_score(y_test, y_pred_logistic),
        accuracy_score(y_test, y_pred_tree),
        accuracy_score(y_test, y_pred_forest),
        accuracy_score(y_test, y_pred_svm),
        accuracy_score(y_test, y_pred_knn)
    ],

    "Precision": [
        precision_score(y_test, y_pred_logistic),
        precision_score(y_test, y_pred_tree),
        precision_score(y_test, y_pred_forest),
        precision_score(y_test, y_pred_svm),
        precision_score(y_test, y_pred_knn)
    ],

    "Recall": [
        recall_score(y_test, y_pred_logistic),
        recall_score(y_test, y_pred_tree),
        recall_score(y_test, y_pred_forest),
        recall_score(y_test, y_pred_svm),
        recall_score(y_test, y_pred_knn)
    ],

    "F1 Score": [
        f1_score(y_test, y_pred_logistic),
        f1_score(y_test, y_pred_tree),
        f1_score(y_test, y_pred_forest),
        f1_score(y_test, y_pred_svm),
        f1_score(y_test, y_pred_knn)
    ]
})

# Sort models from highest to lowest accuracy
comparison_df = comparison_df.sort_values(
    by="Accuracy",
    ascending=False
)

# Show comparison table
print(comparison_df)

# Find best model
best_model = comparison_df.iloc[0]

# Find worst model
worst_model = comparison_df.iloc[-1]

print("\nBest performing model:")
print(best_model["Model"])
print("Accuracy:", best_model["Accuracy"])

print("\nWorst performing model:")
print(worst_model["Model"])
print("Accuracy:", worst_model["Accuracy"])
```
