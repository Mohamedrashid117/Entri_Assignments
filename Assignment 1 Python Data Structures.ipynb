### Topic :List

1. Create a list of 5 random numbers and print the list.
2. Insert 3 new values to the list and print the updated list.
3. Try to use a for loop to print each element in the list.

```python
rand=(3,5,10,20,25)
print(rand)
```

```python
rand=list(rand)
rand.extend([30,40,45])
print(rand)
```

```python
for i in rand:
    print(i)
```
