4.Write a NumPy program that creates a 2D array x of shape (3, 4) and a 1D array y of shape (4,). Subtract y from each row of x using broadcasting. (Score: 2)

```python
print(f"This is {x.ndim}D array")
x=np.array([[1,2,3,4],[5,6,7,8],[8,3,4,5]])
print(x)

print(f"This is {y.ndim}D array")
y = np.array([2,2,2,2])
print(y)
print("Result")
result= x - y
print(result)
```
